#### (2) boxplot — 그룹별 분포 비교 🟢
📌 상자 가운데 선 = **중앙값**, 상자 아래·위 = **Q1(25%)·Q3(75%)**, 수염 끝 = Q1−1.5×IQR ~ Q3+1.5×IQR, 그 밖의 점 = **이상치 후보**.

```
   ●  ← 이상치 후보 (수염 밖)
   ┬  ← 위 수염 끝 (Q3 + 1.5×IQR 이내의 최대값)
 ┌─┴─┐ ← Q3 (75%)
 │───│ ← 중앙값 (50%)
 └─┬─┘ ← Q1 (25%)      IQR = Q3 − Q1 (상자 높이)
   ┴  ← 아래 수염 끝
```

### 그래프 치트시트
| 언제? | 그래프 | 코드 패턴 |
|---|---|---|
| 글자(범주) 값마다 **몇 개?** | countplot | `sns.countplot(data=df, x='컬럼')` |
| 숫자 하나의 **분포** | histplot | `sns.histplot(data=df, x='컬럼', hue='정답', kde=True)` |
| 그룹별 분포 비교 · **이상치** | boxplot | `sns.boxplot(data=df, x='그룹', y='숫자')` |
| 숫자 두 개의 **관계 + 분포** | jointplot | `sns.jointplot(data=df, x='숫자1', y='숫자2')` |
| 숫자 전체의 **상관관계** | heatmap | `sns.heatmap(df.corr(numeric_only=True), annot=True)` |
| 그래프 **여러 개 나란히** | subplots | `fig, axes = plt.subplots(1, 2)` → `ax=axes[0]` → `axes[0].set_title()` |
