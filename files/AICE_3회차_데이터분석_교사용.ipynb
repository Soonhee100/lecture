시험 3·4번에는 countplot·jointplot 말고도 **histplot, boxplot, heatmap**이 언제든 나올 수 있어요. 같은 데이터로 연습해 봐요.

#### (1) histplot — 숫자 컬럼 하나의 분포 🟢
📌 숫자를 구간으로 나눠서 각 구간에 몇 개가 있는지. `kde=True`를 주면 부드러운 곡선이 함께 그려져요.

````python
sns.histplot(data=df, x='Time_Driving', kde=True)
````

#### (2) boxplot — 그룹별 분포 비교 🟢
📌 상자 가운데 선 = **중앙값**, 상자 아래·위 = **Q1(25%)·Q3(75%)**, 수염 끝 = Q1−1.5×IQR ~ Q3+1.5×IQR, 그 밖의 점 = **이상치 후보**.

```
   ●  ← 이상치 후보 (수염 밖)
   ┬  ← 위 수염 끝 (Q3 + 1.5×IQR 이내의 최대값)
 ┌─┴─┐ ← Q3 (75%)
 │───│ ← 중앙값 (50%)
 └─┬─┘ ← Q1 (25%)      IQR = Q3 − Q1 (상자 높이)
   ┴  ← 아래 수염 끝
```

````python
sns.boxplot(data=df, x='Address1', y='Time_Driving')
````

#### (3) heatmap — 숫자 컬럼들끼리의 상관관계 한눈에 보기 ★
📌 `df.corr()`로 상관계수 표를 만들고, 그 표를 색으로 칠한 것이 heatmap이에요.
- **+1에 가까움**: 하나가 커지면 다른 것도 커짐 / **−1에 가까움**: 하나가 커지면 다른 것은 작아짐 / **0 근처**: 관계 거의 없음
- 글자 컬럼이 섞여 있으면 에러가 나므로 `numeric_only=True`를 넣어 숫자 컬럼만 계산해요.

````python
corr = df.corr(numeric_only=True)
plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm')
plt.show()
````

#### 👩‍🏫 해설 — 확장 그래프
- **histplot**: 오른쪽으로 꼬리가 긴 분포(대부분 500~1,200초, 최대 5,010초). 평균(968) > 중앙값(881)인 이유가 이 꼬리 때문.
- **boxplot**: 중앙값이 가장 높은 곳은 **서울특별시(약 1,011초)**, 경기도는 약 788초. 서울은 거리가 같아도 신호가 많아 오래 걸린다는 해석으로 연결 가능. Part C 3번 문항의 정답 근거.
- **heatmap**: Time_Driving과 상관 — Signaltype 0.54, Distance 0.54, **Speed_Per_Hour −0.43**(유일한 음의 상관), 요일·시각·날짜는 거의 0.
  - `annot=True`: 칸 안에 숫자 표시 / `fmt='.2f'`: 소수 둘째 자리 / `cmap`: 색 지도
  - 구버전 pandas(시험 환경)는 `df.corr()`만 써도 경고와 함께 숫자 컬럼만 계산하지만, Colab 최신 pandas는 에러가 나므로 `numeric_only=True`를 습관화.
